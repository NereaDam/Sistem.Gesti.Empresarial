```python
# EJERCICIO 06 - VERSIÓN 1: leyendo el archivo línea a línea con for

class Tienda:
    def __init__(self):
        # Diccionario vacío: nombre del producto -> precio
        self.productos = {}

    # a) Lee el archivo y guarda los productos con su precio como float
    def cargar(self, archivo):
        with open(archivo, "r") as f:
            for linea in f:
                partes = linea.split()          # "teclado 25.50\n" -> ["teclado", "25.50"]
                if len(partes) == 2:            # ignora las líneas en blanco
                    self.productos[partes[0]] = float(partes[1])

    # a) Muestra los productos: nombre en 10 caracteres, precio en 8 con 2 decimales
    def mostrar(self):
        for nombre, precio in self.productos.items():
            print(f"{nombre:10}{precio:8.2f} €")

    # b) Recibe un diccionario {producto: unidades} y escribe el ticket
    def compra(self, pedido):
        subtotal = 0
        for nombre, unidades in pedido.items():
            if nombre in self.productos:                 # ¿existe el producto?
                precio = self.productos[nombre]
                importe = unidades * precio
                subtotal = subtotal + importe
                # nombre 10, unidades 3, precio 7 y importe 8 (con 2 decimales)
                print(f"{nombre:10}{unidades:3} x {precio:7.2f} = {importe:8.2f}")
            else:
                print(f"{nombre} no está en la tienda")   # no se cuenta
        iva = subtotal * 0.21        # 21 %
        total = subtotal + iva
        print(f"{'Subtotal:':11}{subtotal:8.2f} €")
        print(f"{'IVA (21%):':11}{iva:8.2f} €")
        print(f"{'Total:':11}{total:8.2f} €")


tienda = Tienda()
tienda.cargar("productos.dat")
tienda.mostrar()
print()
tienda.compra({"teclado": 2, "raton": 1, "silla": 4})
```

```python
# EJERCICIO 06 - VERSIÓN 2: usando la librería numpy

import numpy as np


class Tienda:
    def __init__(self):
        self.productos = {}

    # a) np.loadtxt lee el archivo entero y lo convierte en una tabla (array)
    # dtype=str: leemos todo como texto; luego pasamos el precio a float
    def cargar(self, archivo):
        datos = np.loadtxt(archivo, dtype=str)    # [["teclado", "25.50"], ["raton", "12.00"], ...]
        for nombre, precio in datos:              # cada fila tiene 2 valores
            self.productos[str(nombre)] = float(precio)

    def mostrar(self):
        for nombre, precio in self.productos.items():
            print(f"{nombre:10}{precio:8.2f} €")

    # b) Guardamos los importes en una lista y los sumamos con numpy
    def compra(self, pedido):
        importes = []
        for nombre, unidades in pedido.items():
            if nombre in self.productos:
                precio = self.productos[nombre]
                importe = unidades * precio
                importes.append(importe)
                print(f"{nombre:10}{unidades:3} x {precio:7.2f} = {importe:8.2f}")
            else:
                print(f"{nombre} no está en la tienda")
        subtotal = np.sum(importes)      # suma de todos los importes (0 si no hay ninguno)
        iva = subtotal * 0.21
        total = subtotal + iva
        print(f"{'Subtotal:':11}{subtotal:8.2f} €")
        print(f"{'IVA (21%):':11}{iva:8.2f} €")
        print(f"{'Total:':11}{total:8.2f} €")


tienda = Tienda()
tienda.cargar("productos.dat")
tienda.mostrar()
print()
tienda.compra({"teclado": 2, "raton": 1, "silla": 4})
```
