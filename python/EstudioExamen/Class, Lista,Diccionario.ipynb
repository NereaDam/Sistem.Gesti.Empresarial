```python
class Perro: #clase perro
    def __init__(self, nombre, edad): #metodo init con las variables a usar, lo que define al perro
        self.nombre = nombre #nombre
        self.edad = edad #edad 
    #Este metodo muestra porque tiene print 
    def ladrar(self): #Metodo que define una accion del perro como ladrar
        print(self.nombre + " dice: guau") #el perro con nombre ladra 

    #Este metodo devuelve algo con el return
    def get_nombre(self):
        return self.nombre

    #este metodo camia el dato guardado, tenemos la edad del perro y como cumple años le sumamos 1
    def cumplir_anios(self):
        self.edad = self.edad + 1

    




    
mi_perro = Perro("Toby", 3) #esto es la creacion de un objeto, en este caso de un perro que se llama tobi y tiene 3 años
print(mi_perro.nombre) #aqui se imporime el nombre de mi perro
mi_perro.ladrar() #Y aqui imprimo lo que hace mi perro, que lo he puesto en mi metodo ladrar  
print("======================")
#Esta es otra forma de decir que tu perro se llama toby y tiene 3 años
toby=Perro("Toby",3)
print("se llama ", toby.get_nombre()) #aqui cogemos el nombre de toby
print("tiene " , toby.edad , "años y va a ser su cumpleaños ") #aqui decimos la edad de tobi
toby.cumplir_anios() #aqui decimos que toby cumplio años 
print("ahora tiene " , toby.edad) # aqui imprimimos la edad de tobi
```

```python
frutas = ["manzana", "pera", "uva"] #cremao la lista
print(frutas[0]) #Imprimimos la posicon 0 de la lista
frutas.append("kiwi") #Añadimos kiwi a la lista
print(len(frutas)) #Nos dice cuantas frutas hay en la lista (solo el numero)

print("===================")

#RECORRER UN FOR...IN
#for fruta in frutas: se lee “para cada fruta en frutas”. Python toma los elementos de uno en uno y en cada vuelta ejecuta lo que está tabulado debajo.

for fruta in frutas:
    print(fruta)

```

```python
#Diccionario: un armario con etiquetas
#En vez de números, cada cosa tiene una etiqueta (la clave) y guarda un valor.
```

```python
precios = {"teclado": 25.5, "raton": 12.0} #Creamos el diccionario
print(precios["teclado"]) #Le decimos que queremos solo el precio del teclado
precios["monitor"] = 149.99 #añade una pareja nueva (o cambia el valor si la clave ya existía).
print("raton" in precios)#pregunta “¿existe esta etiqueta?” y responde True o False
#En raton sale true 
#en silla sale false 
print("silla" in precios)
#Que hacemos para añadir silla al diccionario?
precios["silla"] = 40.0 #Decimos que queremos una silla y que valga 40


print("======SE IMPRIME EL PRODUCTO Y EL PRECIO=========")

#Con un diccionario, .items() te da la etiqueta y el valor a la vez:

#para cada producto y cada precio en precios dame etiqueta y valor
for producto, precio in precios.items():
    print(producto, precio)
print("=======SOLO IMPRIMIMOS PRODUCTO========")
#Aqui solo imprimirmos el producto
for producto in precios:
    print(producto)

print("===========")
for producto in precios.keys():#otra forma de imprimri solo el producto es ponerle keys()
    print(producto)
    
print("======SOLO IMPRIMIRMOS PRECIOS=========")
for precio in precios.values(): #solo nos imprime los valores
    print(precio)
    
```

```python
class Clase: #Definimos la clase
    def __init__(self, nombre): #inicimos el metodo con atrobutos
        self.nombre = nombre
        self.alumnos = [] #Hace que cada objeto tenga su propia lista vacia al crearlo
        #No hay que pasarla como parametro
#alumnos no esta entre el parentesis porque lo que va entre parentesis
#es lo que puedes contruir despues y lo que haces es darle el nomnre y podriamos poner tambien edad
    #Alumnos[] simplemente es la lista donde guardas el nombre
    
    def anadir(self, alumno):#Añadimos alumno, pasandole el alumno
        self.alumnos.append(alumno)

    def mostrar(self):
        for alumno in self.alumnos: #muestra la lista de alumnos, en cada vuelta saca un nombre
            print(alumno)

    def mostrar_lista(self): #metodo que devuelve la lista entera
        print(self.alumnos)




dam1 = Clase("DAM1")
dam1.anadir("Ana")
dam1.anadir("Luis")
dam1.mostrar()
print("=======================")
#para imprimir la lista entera de nombre seria asi:
dam1.mostrar_lista()
```

```python
class Objeto:
    def __init__(self,objeto, precio):
        self.objeto=objeto
        self.precio=precio
        self.objetos ={} #se crea el diccionario

    def anadir(self, objeto,precio): #Añadimos objeto y precio, pasandole el objeto y precio
         self.objetos[objeto]=precio

    def mostrar(self):
        for objeto,precio in self.objetos.items(): #muestra la lista de objetos y precios, en cada vuelta saca un objeto y precio
            print(objeto,precio)

    def mostrar_diccionario(self): #metodo que devuelve la lista entera
        print(self.objetos)



objeto = Objeto("objeto", 0) #crea el objeto
objeto.anadir("Silla",4)
objeto.anadir("Lampara",3)
objeto.mostrar()
print("=======================")
#para imprimir la lista entera de nombre seria asi:
objeto.mostrar_diccionario()
```

```python
class Tienda:  # molde: cada tienda tendrá su diccionario de objetos y precios
    def __init__(self):  # no recibe datos de fuera, empieza vacía
        self.objetos = {}  # diccionario vacío: nombre -> precio
#El init se inicia solo porque despues en añadir ya le damos el objeto y el precio
    def anadir(self, objeto, precio):  # recibe un objeto y su precio
        self.objetos[objeto] = precio  # guarda la pareja en el diccionario

    def mostrar(self):  # muestra cada objeto con su precio
        for objeto, precio in self.objetos.items():  # en cada vuelta: un nombre y su precio
            print(objeto, precio)  # los escribe en una línea

    def mostrar_diccionario(self):  # muestra el diccionario entero de golpe
        print(self.objetos)


tienda = Tienda()  # crea la tienda (sin datos entre paréntesis)
tienda.anadir("Silla", 4)  # añade Silla con precio 4
tienda.anadir("Lampara", 3)  # añade Lampara con precio 3
tienda.mostrar()
print("=======================")
tienda.mostrar_diccionario()
```

```python
class Mochila:

    def __init__(self,nombreCosa):
        self.nombreCosa=nombreCosa
        self.cosas=[]

    def guardar(self, nombreCosa):
        self.cosas.append(nombreCosa)

    def mostrar(self):
        for nombreCosa in self.cosas:
            print(nombreCosa)


mochila=Mochila("nombreCosa")
mochila.guardar("Libro")
mochila.guardar("Bocadillo")
mochila.mostrar()
        
```
