```python
#Leer Archivos 
#"w": escribir en un archivo y borra lo que ya habia (si no tenia nada, no pasa nada)
#"r" : leer un nuevo archivo y no borra lo anterior
#"a" : Añadir al final del archivo y no borra lo anterior
```

```python
#"w"
#Abre el archivo prueba y si no existe lo crea
# as f, solo es el nombre F para poder hablarnde el 
with open("prueba.txt", "w") as f: 
    f.write("Hola\n") #escribe la palabra Hola. El \n es un salto de línea, como pulsar Enter.
    f.write("Adios\n")
#solo con lo de arriba se escribe en el documento pero solo lo ves si abres el documento


#"r" leer el archivo sin abrir la pestaña
with open("prueba.txt", "r") as f:
    print(f.read())
  
```

```python
with open("productos.dat", "w") as f:
    f.write("teclado 25.50\n")
    f.write("raton 12.00\n")
    f.write("monitor 149.99\n")
    f.write("webcam 45.90\n")
```

```python
#"r" leer el archivo sin abrir la pestaña
with open("productos.dat", "r") as f:
    #Cada vuelta del for te da una línea.
    for linea in f:
        #strip() quita los espacios y el \n de los extremos, 
        #y así no salen líneas en blanco de más.
        print(linea.strip())
```

```python

#"a" Añadir cosas al archivo

with open("productos.dat", "a") as f:
    f.write("auriculares 19.99\n")
```

```python
#"r" leermos que se ha añadido
with open("productos.dat", "r") as f:
    for linea in f:
        print(linea.strip())
```

```python
#Borrar una linea concreta 


#leermos el archivo
with open("productos.dat", "r") as f:
#f.readlines() mete las 5 líneas en la lista lineas. 
#Ahora están guardadas en la memoria, no solo en el archivo.
    lineas = f.readlines() 

#Le decimos que queremos escribir con W pero en realidad es borrar
with open("productos.dat", "w") as f:
    for linea in lineas: #recorre las líneas una a una.
        #linea.startswith("raton") pregunta “¿esta línea empieza por raton?” y responde True o False.
        #not le da la vuelta: if not ... significa “si no empieza por raton”.
        if not linea.startswith("raton"): 
            #Ecribe solo las lineas que NO tengan raton
            f.write(linea) #escribe la línea. Como ya trae su \n, no hay que añadir nada.


#volvemos a leer imprimiendo
with open("productos.dat", "r") as f:
    for linea in f:
        print(linea.strip())
            
```

```python
with open("notas.txt", "w") as f:
    f.write("Ana 8.5\n")
    f.write("Luis 6.0\n")
    f.write("Marta 9.25\n")
```

```python
class Notas: #clase notas
    def __init__(self):
        self.notas = {} #creamos un diccionario de notas

    def cargar(self, nombreArchivo):
        with open(nombreArchivo, "r") as f: #abre el archivo para leer
            for linea in f: #lee una a una las lineas del archivo
                partes = linea.split() #corta la linea por epacios
                nombre = partes[0] #la primera parte 
                nota = float(partes[1]) #la segunda parte que son los numeros y es un float
                self.notas[nombre] = nota # guarda el nombre y la notae n el dicionario (guarda 'Ana': 8.5 en el diccionario.)
#no se meten como parametros nombre y nota porque ya los tenemos arriba
#aqui no se rellenan, solo se muestran
    def mostrar(self): 
        for nombre,nota in self.notas.items():
            print(nombre,nota)

    def media(self):
        suma=sum(self.notas.values())
        cantidad=len(self.notas)
        return suma/cantidad
    


    
registro = Notas() #crea el objeto registro
registro.cargar("notas.txt") #registro llama a cargar y este lee el archivo
print(registro.notas) #se imprime el diccionario notas.txt

print("=======================")

registro.mostrar()
print("Hay", len(registro.notas), "notas")

#ESTA ES UNA FORMA DE SACAR LAS NOTAS
print("Las notas son", list(registro.notas.values()))

print("=============================")

#ESTA ES OTRA FORMA DE SACAR LA LISTA DE LAS NOTAS 
textos = []
for nota in registro.notas.values():
    textos.append(str(nota))

print("Las notas son", ", ".join(textos))

print("La media es", registro.media())
```

```python
#FORMATO DE SALIDA
#Sirve para que lo que escribes salga alineado y con los decimales que quieras.
```

```python
#la f-string
#Pones una f delante de las comillas, y lo que escribas entre llaves { } se sustituye por el valor de esa variable:

nombre = "teclado"
precio = 25.5
print(f"Producto: {nombre}")
```

```python
#ocupar un ancho fijo
#Después de la variable pones : y un número. 
#Es el número de caracteres que ocupará, rellenando con espacios. 
#Para verlo mejor pongo barras | a los lados:


print(f"|{nombre:10}|")
#TECLADO tiene 7 letras asique se añaden 3 espacios
```

```python
#Decimales
#.2f significa “número decimal con 2 decimales”:

print(f"|{precio:.2f}|")
```

```python
#si juntas un ancho y los decimales, 8.2f es “ancho 8, con 2 decimales”:

print(f"|{precio:8.2f}|")

#quí los espacios aparecen a la izquierda, porque los números se pegan a la derecha.
```

```python
#JUNTANDOLO TODO

print(f"{nombre:10}{precio:8.2f} €")
#NOMBRE en 10, PRECIO en 8 con 2 decimales y despues €
```
