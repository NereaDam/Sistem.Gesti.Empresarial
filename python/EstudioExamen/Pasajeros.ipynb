```python
# EJERCICIO 05 - VERSIÓN 2: con list comprehension y un método __str__

lista = [["Mujer","Antonella","Rodriguez"],
["Mujer","Isabel","Gutierrez"],
["Mujer","Luna","Garcia"],
["Mujer","Luciana","Gonzalez"],
["Mujer","Ana","Martinez"],
["Mujer","Sofía","Medina"],
["Hombre","Mario","Herrera"],
["Mujer","Francesca","Sanchez"],
["Hombre","Alejandro","Sanchez"],
["Mujer","Alejandra","Romero"],
["Mujer","Allison","Castro"],
["Hombre","Isaac","Vazquez"],
["Hombre","Carlos","Gonzalez"],
["Mujer","Renata","Morrell"],
["Mujer","Laura","Sanchez"],
["Hombre","Sebastián","Molina"],
["Mujer","Abigail","Lopez"],
["Mujer","Andrea","Ramos"],
["Mujer","María","Delgado"],
["Mujer","Ivanna","Serrano"],
["Mujer","Alejandra","Medina"],
["Mujer","Violeta","Ortega"],
["Mujer","Ashley","Rodriguez"],
["Mujer","Julia","Alvarez"],
["Mujer","Miranda","Cruz"],
["Mujer","Nicole","Gil"],
["Mujer","Rafaela","Herrera"],
["Hombre","Samuel","Mendez"],
["Hombre","Kevin","Castro"],
["Mujer","Lucía","Santos"],
["Hombre","LiaHombre","Gil"],
["Mujer","Allison","Jimenez"]]


class Persona:
    def __init__(self, nacimiento, sexo, nombre, apellido, cara="Alegre"):
        self.nacimiento = nacimiento
        self.sexo = sexo
        self.nombre = nombre
        self.apellido = apellido
        self.cara = cara

    # Define cómo se escribe una Persona con print(persona)
    def __str__(self):
        return f"El pasajero {self.nombre} tiene cara {self.cara}"


# List comprehension: crea la lista entera en una sola línea
# Se lee: "una Persona por cada (sexo, nombre, apellido) de la lista"
pasajero = [Persona(2000, sexo, nombre, apellido) for sexo, nombre, apellido in lista]

# CAMBIA ESTOS TRES VALORES POR LOS TUYOS:
dia_cumple = 26
mi_nombre = "Daniel"
mi_sexo = "Hombre"

# Asignamos varios atributos seguidos al pasajero elegido
elegido = pasajero[dia_cumple - 1]
elegido.nombre, elegido.sexo, elegido.cara = mi_nombre, mi_sexo, "DE CUMPLEAÑOS"

# print(p) usa automáticamente el __str__ de Persona
for p in pasajero:
    print(p)
```
