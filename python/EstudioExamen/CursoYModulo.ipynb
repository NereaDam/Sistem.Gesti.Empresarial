```python
# EJERCICIO 04 - VERSIÓN 1: método info() que recorre los módulos con un for

DAM1 = [["0484", "Bases de datos", 6],
        ["0487", "Entornos de desarrollo", 3],
        ["0493", "Formación y orientación laboral", 3],
        ["0373", "Lenguaje de marcas y sistemas de gestión de información", 4],
        ["0485", "Programación", 8],
        ["0483", "Sistemas informáticos", 6]]
DAM2 = [["0486", "Acceso a datos", 6],
        ["0488", "Desarrollo de interfaces", 6],
        ["0494", "Empresa e iniciativa emprendedora", 3],
        ["9009", "Inglés técnico para grado superior", 2],
        ["0490", "Programación de servicios y procesos", 4],
        ["0489", "Programación multimedia y dispositivos móviles", 4],
        ["0491", "Sistemas de gestión empresarial", 5]]


# Un módulo: código, nombre y horas
class Modulo:
    def __init__(self, ID, nombre, horas):
        self.ID = ID
        self.nombre = nombre
        self.horas = horas


# Un curso: año, nombre y una lista de módulos que empieza vacía
class Curso:
    def __init__(self, año, nombre):
        self.año = año
        self.nombre = nombre
        self.modulos = []        # lista vacía: se rellena con .append(...) desde fuera

    # Escribe el curso y, debajo, cada módulo con dos espacios de sangría
    def info(self):
        print("Curso", self.año, self.nombre)
        for m in self.modulos:
            print(f"  {m.ID} {m.nombre} ({m.horas})")


cursos = [Curso(1, "DAM"), Curso(2, "DAM")]

for i in DAM1:
    cursos[0].modulos.append(Modulo(i[0], i[1], i[2]))

for i in DAM2:
    cursos[1].modulos.append(Modulo(i[0], i[1], i[2]))

for i in cursos:
    i.info()
```

```python
# EJERCICIO 04 - VERSIÓN 2: con __str__, un método cargar() y total de horas

DAM1 = [["0484", "Bases de datos", 6],
        ["0487", "Entornos de desarrollo", 3],
        ["0493", "Formación y orientación laboral", 3],
        ["0373", "Lenguaje de marcas y sistemas de gestión de información", 4],
        ["0485", "Programación", 8],
        ["0483", "Sistemas informáticos", 6]]
DAM2 = [["0486", "Acceso a datos", 6],
        ["0488", "Desarrollo de interfaces", 6],
        ["0494", "Empresa e iniciativa emprendedora", 3],
        ["9009", "Inglés técnico para grado superior", 2],
        ["0490", "Programación de servicios y procesos", 4],
        ["0489", "Programación multimedia y dispositivos móviles", 4],
        ["0491", "Sistemas de gestión empresarial", 5]]


class Modulo:
    def __init__(self, ID, nombre, horas):
        self.ID = ID
        self.nombre = nombre
        self.horas = horas

    # __str__ define cómo se convierte el objeto en texto cuando hacemos print(modulo)
    def __str__(self):
        return f"{self.ID} {self.nombre} ({self.horas})"


class Curso:
    def __init__(self, año, nombre):
        self.año = año
        self.nombre = nombre
        self.modulos = []

    # Recibe una lista como DAM1 y crea un Modulo por cada elemento
    def cargar(self, lista):
        for ID, nombre, horas in lista:    # desempaqueta cada sublista en 3 variables
            self.modulos.append(Modulo(ID, nombre, horas))

    # Suma las horas de todos los módulos
    def total_horas(self):
        total = 0
        for m in self.modulos:
            total += m.horas
        return total

    # __str__ del curso: una cabecera y un módulo por línea con sangría
    def __str__(self):
        texto = f"Curso {self.año} {self.nombre}"
        for m in self.modulos:
            texto += f"\n  {m}"
        return texto

    # info() sigue existiendo, como pide el enunciado: solo imprime el texto del __str__
    def info(self):
        print(self)


cursos = [Curso(1, "DAM"), Curso(2, "DAM")]
cursos[0].cargar(DAM1)
cursos[1].cargar(DAM2)

for i in cursos:
    i.info()

print("Horas de 1º:", cursos[0].total_horas())
print("Horas de 2º:", cursos[1].total_horas())
```
