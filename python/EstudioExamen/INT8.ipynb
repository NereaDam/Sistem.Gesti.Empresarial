```python
#Interpreto “rellenar con ceros” como un número de 8 caracteres en total
#(501 → 00000501). Así 9999999 cabe justo y el negativo lleva el signo
#dentro de los 8.
#Con zfill()
```

```python
# EJERCICIO 01 - VERSIÓN 1: usando zfill()

class int8:
    # __init__ se ejecuta solo al crear el objeto: int8("501")
    def __init__(self, numero):
        # int() convierte "501" (texto) o 501 (número) en un número entero
        self.numero = int(numero)

    # Método auxiliar: dice si un valor se sale del rango permitido
    def _fuera_de_rango(self, valor):
        return valor > 9999999 or valor < -9999999

    # Devuelve el número con ceros a la izquierda hasta tener 8 caracteres
    def getint8(self):
        if self._fuera_de_rango(self.numero):
            return "NAN"
        # str() pasa el número a texto; zfill(8) rellena con ceros a la izquierda
        # (si el número es negativo, zfill deja el signo - delante)
        return str(self.numero).zfill(8)

    # Suma este número con otro objeto int8 y devuelve el resultado con 8 caracteres
    def sum(self, otro):
        total = self.numero + otro.numero
        if self._fuera_de_rango(total):
            return "NAN"
        return str(total).zfill(8)


a = int8("501")
b = int8("101")
print(a.getint8())   # 00000501
print(a.sum(b))      # 00000602
nmax = int8(9999999)
print(a.sum(nmax))   # NAN
print(int8("-501").getint8())  # -0000501
```

```python
# EJERCICIO 01 - VERSIÓN 2: usando f-string con formato :08d

class int8:
    def __init__(self, numero):
        # int() convierte el texto o número que nos den en un entero
        self.numero = int(numero)

    # Método auxiliar que da formato a cualquier valor (así no repetimos código)
    def _formato(self, valor):
        # Si se sale del rango permitido, devolvemos "NAN"
        if valor > 9999999 or valor < -9999999:
            return "NAN"
        # :08d = entero (d), ocupando 8 caracteres (8), rellenando con ceros (0)
        return f"{valor:08d}"

    def getint8(self):
        return self._formato(self.numero)

    def sum(self, otro):
        return self._formato(self.numero + otro.numero)


a = int8("501")
b = int8("101")
print(a.getint8())   # 00000501
print(a.sum(b))      # 00000602
nmax = int8(9999999)
print(a.sum(nmax))   # NAN
print(int8("-501").getint8())  # -0000501
```
