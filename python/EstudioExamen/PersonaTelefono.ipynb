```python
# EJERCICIO 03 - VERSIÓN 1: la persona TIENE un teléfono (composición)

class Telefono:
    def __init__(self, numero):
        # Guardamos el número como texto para poder mirar su primer dígito
        self.numero = str(numero)
        self.tipo = self._calcular_tipo()

    # Mira la primera cifra: 8 o 9 -> fijo; 6 o 7 -> móvil
    def _calcular_tipo(self):
        if self.numero[0] in "89":
            return "fijo"
        if self.numero[0] in "67":
            return "móvil"
        return "desconocido"

    # Cambia el número solo si tiene exactamente 9 cifras (y son todo dígitos)
    # Devuelve True si lo ha cambiado y False si no
    def cambiar(self, nuevo):
        nuevo = str(nuevo)
        if len(nuevo) != 9 or not nuevo.isdigit():
            return False
        self.numero = nuevo
        self.tipo = self._calcular_tipo()   # recalculamos si es fijo o móvil
        return True

    def info(self):
        return f"Mi teléfono es un {self.tipo} con número {self.numero}"


class Persona_tlf:
    def __init__(self, nacimiento, sexo, nombre, apellido, telefono):
        self.nacimiento = nacimiento
        self.sexo = sexo
        self.nombre = nombre
        self.apellido = apellido
        # La persona tiene un objeto Telefono dentro
        self.telefono = Telefono(telefono)

    def get_tlf(self):
        print(self.telefono.numero)

    def tlf_info(self):
        print(self.telefono.info())

    def cambiar_tlf(self, nuevo):
        if self.telefono.cambiar(nuevo):
            print("Número cambiado,", self.telefono.info())
        else:
            print("Número no cambiado, necesitamos un número de 9 cifras")


persona_tlf01 = Persona_tlf(1979, "Hombre", "Manuel", "López", "987654321")
persona_tlf01.get_tlf()              # 987654321
persona_tlf01.tlf_info()             # Mi teléfono es un fijo con número 987654321
persona_tlf01.cambiar_tlf(1235)      # Número no cambiado, necesitamos un número de 9 cifras
persona_tlf01.cambiar_tlf(654987321) # Número cambiado, Mi teléfono es un móvil con número 654987321
```

```python
# EJERCICIO 03 - VERSIÓN 2: herencia (Persona_tlf ES una Persona) y diccionario de tipos

# Diccionario: primera cifra -> tipo de teléfono
TIPOS = {"9": "fijo", "8": "fijo", "6": "móvil", "7": "móvil"}


class Telefono:
    def __init__(self, numero):
        self.numero = str(numero)
        # .get(clave, valor_por_defecto) busca la primera cifra en el diccionario
        self.tipo = TIPOS.get(self.numero[0], "desconocido")

    def cambiar(self, nuevo):
        nuevo = str(nuevo)
        if len(nuevo) != 9 or not nuevo.isdigit():
            return False
        self.numero = nuevo
        self.tipo = TIPOS.get(nuevo[0], "desconocido")
        return True

    def info(self):
        return f"Mi teléfono es un {self.tipo} con número {self.numero}"


# Clase base: una persona normal, sin teléfono
class Persona:
    def __init__(self, nacimiento, sexo, nombre, apellido):
        self.nacimiento = nacimiento
        self.sexo = sexo
        self.nombre = nombre
        self.apellido = apellido


# Persona_tlf HEREDA de Persona: ya tiene nacimiento, sexo, nombre y apellido
class Persona_tlf(Persona):
    def __init__(self, nacimiento, sexo, nombre, apellido, telefono):
        # super().__init__ llama al __init__ de Persona para guardar los 4 primeros datos
        super().__init__(nacimiento, sexo, nombre, apellido)
        self.telefono = Telefono(telefono)

    def get_tlf(self):
        print(self.telefono.numero)

    def tlf_info(self):
        print(self.telefono.info())

    def cambiar_tlf(self, nuevo):
        if self.telefono.cambiar(nuevo):
            print("Número cambiado,", self.telefono.info())
        else:
            print("Número no cambiado, necesitamos un número de 9 cifras")


persona_tlf01 = Persona_tlf(1979, "Hombre", "Manuel", "López", "987654321")
persona_tlf01.get_tlf()
persona_tlf01.tlf_info()
persona_tlf01.cambiar_tlf(1235)
persona_tlf01.cambiar_tlf(654987321)
```
