```python
#El enunciado dice que el titular “es una persona”, 
#así que creo una clase Persona. 
#La cantidad no tiene setter, porque solo debe cambiar con ingresar y retirar.
```

```python
#Versión 1: métodos get_... y set_...
```

```python
# EJERCICIO 02 - VERSIÓN 1: getters y setters escritos como métodos (get_..., set_...)

# El titular es una persona, así que creamos una clase Persona sencilla
class Persona:
    def __init__(self, nombre, apellidos=""):
        self.nombre = nombre
        self.apellidos = apellidos

    def nombre_completo(self):
        # strip() quita el espacio sobrante si no hay apellidos
        return (self.nombre + " " + self.apellidos).strip()


class Cuenta:
    # titular es obligatorio (no tiene valor por defecto)
    # cantidad es opcional (si no la pasan, vale 0.0)
    def __init__(self, titular, cantidad=0.0):
        # Los dos guiones bajos __ hacen el atributo "privado":
        # desde fuera no se puede hacer cuenta.__cantidad = 5
        self.__titular = titular
        self.__cantidad = float(cantidad)

    # Getter y setter del titular
    def get_titular(self):
        return self.__titular

    def set_titular(self, titular):
        self.__titular = titular

    # Getter de la cantidad. NO hay setter: la cantidad solo cambia
    # con ingresar() y retirar(), como pide el enunciado
    def get_cantidad(self):
        return self.__cantidad

    # Muestra los datos de la cuenta
    def mostrar(self):
        print("Titular:", self.__titular.nombre_completo())
        print(f"Cantidad: {self.__cantidad:.2f} €")

    # Si la cantidad es negativa no se hace nada
    def ingresar(self, cantidad):
        if cantidad > 0:
            self.__cantidad = self.__cantidad + cantidad

    # Se resta la cantidad; la cuenta puede quedar en números rojos (negativa)
    # (añadimos "if cantidad > 0" para que retirar un número negativo no sume dinero)
    def retirar(self, cantidad):
        if cantidad > 0:
            self.__cantidad = self.__cantidad - cantidad


persona = Persona("Ana", "López")
cuenta = Cuenta(persona, 100)
cuenta.mostrar()          # 100.00
cuenta.ingresar(50)       # 150.00
cuenta.ingresar(-20)      # no hace nada
cuenta.retirar(200)       # -50.00 (números rojos)
cuenta.mostrar()

cuenta2 = Cuenta(Persona("Luis"))   # sin cantidad: empieza en 0
cuenta2.mostrar()
cuenta2.set_titular(Persona("Marta"))
print(cuenta2.get_titular().nombre_completo(), cuenta2.get_cantidad())
```

```python
# EJERCICIO 02 - VERSIÓN 2: getters y setters con @property (la forma "pythónica")

class Persona:
    def __init__(self, nombre, apellidos=""):
        self.nombre = nombre
        self.apellidos = apellidos

    def nombre_completo(self):
        return (self.nombre + " " + self.apellidos).strip()


class Cuenta:
    def __init__(self, titular, cantidad=0.0):
        # Un guion bajo _ es la costumbre para decir "no tocar desde fuera"
        self._titular = titular
        self._cantidad = float(cantidad)

    # @property convierte el método en un atributo de solo lectura: cuenta.titular
    @property
    def titular(self):
        return self._titular

    # @titular.setter permite asignar: cuenta.titular = Persona("Marta")
    @titular.setter
    def titular(self, valor):
        self._titular = valor

    # cantidad solo tiene getter, SIN setter:
    # si alguien escribe cuenta.cantidad = 5, Python da un error
    @property
    def cantidad(self):
        return self._cantidad

    def mostrar(self):
        print("Titular:", self._titular.nombre_completo())
        print(f"Cantidad: {self._cantidad:.2f} €")

    def ingresar(self, cantidad):
        # Si es negativa no se hace nada
        if cantidad > 0:
            self._cantidad += cantidad

    def retirar(self, cantidad):
        # Puede quedar en números rojos
        if cantidad > 0:
            self._cantidad -= cantidad


cuenta = Cuenta(Persona("Ana", "López"), 100)
cuenta.mostrar()
cuenta.ingresar(50)
cuenta.ingresar(-20)
cuenta.retirar(200)
cuenta.mostrar()

cuenta.titular = Persona("Marta")    # usa el setter del titular
print(cuenta.titular.nombre_completo(), cuenta.cantidad)

# Comprobamos que la cantidad no se puede modificar directamente
try:
    cuenta.cantidad = 5000
except AttributeError:
    print("No se puede modificar la cantidad directamente")
```
