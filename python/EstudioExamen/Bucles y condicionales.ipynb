```python
#posición en una lista:
class Mochila:
    def __init__(self):
        self.cosas = []

    def guardar(self, cosa):
        self.cosas.append(cosa)

    def mostrar_numerado(self):
#len(self.cosas) es 2 cosas en la mochila (libro y bocadillo)
        for i in range(len(self.cosas)):
            #i+1, empieza desde 1. y self.cosas[i], las posiciones de las cosas en la lista
            print(i + 1 , self.cosas[i])

mochila = Mochila() #creamos solo mochila, no podemos nada en el parentesis porque init solo tiene la lista, no tiene parametros 
mochila.guardar("Libro")
mochila.guardar("Bocadillo")
mochila.mostrar_numerado()

```

```python
#while: repetir mientras algo sea cierto

numero = 1
while numero <= 3:
    print(numero)
    numero = numero + 1 #se le suma 1 en cada vuelta hasta que llega a 3 
```

```python
#if, elif y else: decidir
#Para juntar condiciones usas and (las dos tienen que ser ciertas) y or (basta una).

nota = 7
if nota >= 9:
    print("Sobresaliente")
elif nota >= 5:
    print("Aprobado")
else:
    print("Suspenso")
```

```python
#Un if dentro de un método

class Hucha:
    def __init__(self, cantidad=0):
        self.cantidad = cantidad

    def meter(self, euros):
        if euros > 0:
            self.cantidad = self.cantidad + euros
        else:
            print(euros,"el numero insertado es negativo, vuelva a insertar dinero")

    def sacar(self, euros):
        if euros > 0 and euros <= self.cantidad: #puede sacar dinero si € es mayor a 0 y es menos o igual a la cantidad de la hucha 
            self.cantidad=self.cantidad-euros
        else:
            print("no puede sacar", euros)
            

hucha = Hucha()
hucha.meter(10)
hucha.meter(-5)
print(hucha.cantidad)
hucha.sacar(3)
print(hucha.cantidad)
hucha.sacar(100)
```
