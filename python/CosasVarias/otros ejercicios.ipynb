```python
# EJERCICIOS TÍPICOS DE EXAMEN

# ----- 1. FACTORIAL: 5! = 5 * 4 * 3 * 2 * 1 = 120 -----
# Versión con bucle
def factorial(n):
    resultado = 1
    for i in range(2, n + 1):       # multiplicamos 2 * 3 * ... * n
        resultado *= i              # resultado = resultado * i
    return resultado


# Versión recursiva
def factorial_rec(n):
    if n <= 1:                      # caso base
        return 1
    return n * factorial_rec(n - 1)


print(factorial(5), factorial_rec(5))     # 120 120


# ----- 2. PALÍNDROMO: se lee igual del derecho que del revés -----
def es_palindromo(texto):
    texto = texto.lower().replace(" ", "")   # minúsculas y sin espacios
    return texto == texto[::-1]              # [::-1] da el texto al revés


print(es_palindromo("Anita lava la tina"))   # True
print(es_palindromo("Hola"))                 # False


# ----- 3. FIZZBUZZ: del 1 al 15, múltiplos de 3 -> Fizz, de 5 -> Buzz, de ambos -> FizzBuzz -----
def fizzbuzz(n):
    for i in range(1, n + 1):
        if i % 15 == 0:             # primero el caso de los dos (si no, nunca se llegaría a él)
            print("FizzBuzz")
        elif i % 3 == 0:
            print("Fizz")
        elif i % 5 == 0:
            print("Buzz")
        else:
            print(i)


fizzbuzz(15)


# ----- 4. MÁXIMO COMÚN DIVISOR (algoritmo de Euclides) -----
def mcd(a, b):
    while b != 0:
        a, b = b, a % b             # sustituye (a, b) por (b, resto de a entre b)
    return a


print(mcd(48, 18))                  # 6


# ----- 5. CONTAR VOCALES -----
def contar_vocales(texto):
    contador = 0
    for letra in texto.lower():
        if letra in "aeiouáéíóú":   # "in" mira si la letra está dentro del texto
            contador += 1
    return contador


print(contar_vocales("Programación"))    # 5


# ----- 6. CONTAR PALABRAS con un diccionario -----
def contar_palabras(frase):
    conteo = {}
    for palabra in frase.lower().split():        # split() corta la frase en palabras
        conteo[palabra] = conteo.get(palabra, 0) + 1   # suma 1 (si no existe, empieza en 0)
    return conteo


print(contar_palabras("a b a c b a"))            # {'a': 3, 'b': 2, 'c': 1}


# ----- 7. ORDENAR UNA LISTA (método de la burbuja) -----
def burbuja(lista):
    lista = lista.copy()                         # copia para no modificar la original
    n = len(lista)
    for i in range(n - 1):                       # tantas pasadas como elementos - 1
        for j in range(n - 1 - i):               # compara parejas vecinas
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]    # las intercambia
    return lista


print(burbuja([5, 2, 9, 1, 7]))                  # [1, 2, 5, 7, 9]


# ----- 8. SUMA DE LOS DÍGITOS DE UN NÚMERO -----
# Versión con texto
def suma_digitos(n):
    suma = 0
    for digito in str(n):           # recorremos el número como texto
        suma += int(digito)
    return suma


# Versión con matemáticas: % 10 da la última cifra, // 10 la quita
def suma_digitos_mate(n):
    suma = 0
    while n > 0:
        suma += n % 10              # última cifra
        n //= 10                    # quitamos la última cifra
    return suma


print(suma_digitos(1234), suma_digitos_mate(1234))   # 10 10


# ----- 9. DECIMAL A BINARIO -----
def decimal_a_binario(n):
    if n == 0:
        return "0"
    binario = ""
    while n > 0:
        binario = str(n % 2) + binario      # el resto de dividir entre 2 va por delante
        n //= 2
    return binario


print(decimal_a_binario(10))        # 1010
print(bin(10))                      # 0b1010   (bin() de Python hace lo mismo)


# ----- 10. NÚMERO MAYOR Y MENOR SIN USAR max() NI min() -----
def mayor_menor(lista):
    mayor = lista[0]
    menor = lista[0]
    for numero in lista:
        if numero > mayor:
            mayor = numero
        if numero < menor:
            menor = numero
    return mayor, menor             # devuelve una tupla


print(mayor_menor([4, 9, 1, 7]))    # (9, 1)
```
