```python
# TUPLAS
# Una tupla es como una lista, pero NO se puede modificar (es inmutable).
# Se escribe con paréntesis ( ) y los elementos separados por comas.

tupla = (10, 20, 30)
print(tupla[0])            # 10         primer elemento (se empieza en 0, como en las listas)
print(tupla[-1])           # 30         el último
print(tupla[0:2])          # (10, 20)   trozo (el final no se incluye)
print(len(tupla))          # 3          cuántos elementos tiene
print(tupla.count(10))     # 1          cuántas veces aparece el 10
print(tupla.index(20))     # 1          en qué posición está el 20

# No se puede modificar:
try:
    tupla[0] = 5
except TypeError:
    print("Las tuplas no se pueden modificar")

# Tupla de UN solo elemento: hay que poner una coma
print(type((5)))           # <class 'int'>    sin coma NO es tupla
print(type((5,)))          # <class 'tuple'>  con coma sí

# Se pueden recorrer con for, igual que una lista
for valor in tupla:
    print(valor)

# DESEMPAQUETADO: repartir los elementos en variables
x, y, z = tupla
print(x, y, z)             # 10 20 30

# Intercambiar dos variables (usa una tupla por debajo)
a, b = 1, 2
a, b = b, a
print(a, b)                # 2 1

# Convertir entre lista y tupla
lista = list(tupla)        # tupla -> lista (la lista sí se puede modificar)
lista.append(40)
tupla2 = tuple(lista)      # lista -> tupla
print(tupla2)              # (10, 20, 30, 40)

# Una función puede devolver varios valores: en realidad devuelve una tupla
def min_max(numeros):
    return min(numeros), max(numeros)


minimo, maximo = min_max([4, 9, 1, 7])
print(minimo, maximo)      # 1 9

# Las tuplas sirven como CLAVES de un diccionario (las listas no)
puntos = {(0, 0): "origen", (1, 2): "punto A"}
print(puntos[(1, 2)])      # punto A

# Lista de tuplas + desempaquetado en el for
alumnos = [("Ana", 8.5), ("Luis", 4.0), ("Marta", 9.25)]
for nombre, nota in alumnos:
    print(nombre, nota)

# zip junta dos listas en tuplas
nombres = ["Ana", "Luis"]
notas = [8.5, 4.0]
for nombre, nota in zip(nombres, notas):
    print(nombre, "->", nota)


# ----- EJERCICIO: sacar el alumno con mejor nota de una lista de tuplas -----
mejor = alumnos[0]                 # empezamos suponiendo que el primero es el mejor
for alumno in alumnos:
    if alumno[1] > mejor[1]:       # alumno[1] es la nota (posición 1 de la tupla)
        mejor = alumno
print("Mejor nota:", mejor[0], mejor[1])     # Mejor nota: Marta 9.25
```
