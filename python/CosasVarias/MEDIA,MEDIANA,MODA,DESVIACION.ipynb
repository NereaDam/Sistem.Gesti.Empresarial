```python
# MEDIA, MEDIANA, MODA Y DESVIACIÓN ESTÁNDAR
# Datos de ejemplo:
datos = [2, 4, 4, 4, 5, 5, 7, 9]

# =====================================================
# VERSIÓN 1: hecho a mano (sin librerías)
# =====================================================

# Media = suma de todos / cuántos hay
def media(lista):
    return sum(lista) / len(lista)


# Mediana = valor del medio una vez ordenados
def mediana(lista):
    ordenados = sorted(lista)         # sorted() devuelve una copia ordenada (no toca la original)
    n = len(ordenados)
    medio = n // 2                    # posición del medio (división entera)
    if n % 2 == 1:                    # cantidad impar: hay un valor justo en el medio
        return ordenados[medio]
    # cantidad par: media de los dos valores centrales
    return (ordenados[medio - 1] + ordenados[medio]) / 2


# Moda = valor que más se repite (puede haber varias, por eso devuelve una lista)
def moda(lista):
    conteo = {}                                   # diccionario: valor -> veces que aparece
    for valor in lista:
        conteo[valor] = conteo.get(valor, 0) + 1  # get(valor, 0): si no existe, empieza en 0
    maximo = max(conteo.values())                 # el mayor número de repeticiones
    return [valor for valor, veces in conteo.items() if veces == maximo]


# Desviación estándar = raíz de la media de los cuadrados de las diferencias con la media
# muestral=False -> poblacional (divide entre n);  muestral=True -> divide entre n-1
def desviacion(lista, muestral=False):
    m = media(lista)
    suma_cuadrados = 0
    for valor in lista:
        suma_cuadrados += (valor - m) ** 2        # ** 2 = elevar al cuadrado
    divisor = len(lista) - 1 if muestral else len(lista)
    return (suma_cuadrados / divisor) ** 0.5      # ** 0.5 = raíz cuadrada


print("--- Versión 1: a mano ---")
print("Media:", media(datos))                       # 5.0
print("Mediana:", mediana(datos))                   # 4.5
print("Moda:", moda(datos))                         # [4]
print("Desviación:", desviacion(datos))             # 2.0
print("Desviación muestral:", desviacion(datos, True))   # 2.138089935299395

# =====================================================
# VERSIÓN 2: con numpy
# =====================================================
import numpy as np

arr = np.array(datos)
print("--- Versión 2: numpy ---")
print("Media:", np.mean(arr))                       # 5.0
print("Mediana:", np.median(arr))                   # 4.5
# numpy no tiene moda: unique con return_counts da cada valor y cuántas veces sale
valores, veces = np.unique(arr, return_counts=True)
print("Moda:", valores[veces == veces.max()])       # [4]
print("Desviación:", np.std(arr))                   # 2.0   (poblacional por defecto)
print("Desviación muestral:", np.std(arr, ddof=1))  # 2.138089935299395   (ddof=1 -> n-1)

# =====================================================
# VERSIÓN 3: con el módulo statistics (viene con Python)
# =====================================================
import statistics

print("--- Versión 3: statistics ---")
print("Media:", statistics.mean(datos))             # 5
print("Mediana:", statistics.median(datos))         # 4.5
print("Moda:", statistics.mode(datos))              # 4   (si hay varias, devuelve la primera)
print("Modas:", statistics.multimode(datos))        # [4]  (lista con todas las modas)
print("Desviación:", statistics.pstdev(datos))      # 2.0   poblacional
print("Desviación muestral:", statistics.stdev(datos))   # 2.138089935299395
```
