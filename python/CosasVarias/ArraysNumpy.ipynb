```python
# ARRAYS con numpy
# Un array de numpy es como una lista, pero todos los elementos son del mismo tipo
# y las operaciones se aplican a todos a la vez (sin hacer bucles).

import numpy as np

a = np.array([1, 2, 3, 4, 5])      # crear un array a partir de una lista
print(a)                           # [1 2 3 4 5]
print(a * 2)                       # [ 2  4  6  8 10]   multiplica TODOS
print(a + a)                       # [ 2  4  6  8 10]   suma elemento a elemento
print(a[0], a[-1], a[1:4])         # 1 5 [2 3 4]        índices y trozos, como en listas

# Funciones útiles
print(a.sum())                     # 15
print(a.mean())                    # 3.0   media
print(a.max(), a.min())            # 5 1
print(len(a))                      # 5

# Filtrar con una condición
print(a[a > 2])                    # [3 4 5]   solo los mayores que 2

# Crear arrays de forma rápida
print(np.arange(0, 10, 2))         # [0 2 4 6 8]    como range()
print(np.zeros(3))                 # [0. 0. 0.]
print(np.ones(3))                  # [1. 1. 1.]
print(np.linspace(0, 1, 5))        # [0.   0.25 0.5  0.75 1.  ]   5 números repartidos entre 0 y 1

# Arrays de 2 dimensiones (matrices): una lista de listas
m = np.array([[1, 2, 3],
              [4, 5, 6]])
print(m.shape)                     # (2, 3)   2 filas y 3 columnas
print(m[1, 2])                     # 6        fila 1, columna 2
print(m[0])                        # [1 2 3]  la fila 0
print(m[:, 0])                     # [1 4]    la columna 0 (: = todas las filas)
print(m.sum(axis=0))               # [5 7 9]  suma de cada columna
print(m.sum(axis=1))               # [ 6 15]  suma de cada fila
print(m.reshape(3, 2))             # cambia la forma a 3 filas y 2 columnas

# Diferencia con una lista: lista * 2 repite, array * 2 multiplica
print([1, 2] * 2)                  # [1, 2, 1, 2]
print(np.array([1, 2]) * 2)        # [2 4]

# Arrays con el módulo "array" que viene con Python (menos usado)
from array import array
b = array("i", [1, 2, 3])          # "i" = todos los elementos son enteros
b.append(4)
print(b)                           # array('i', [1, 2, 3, 4])


# ----- EJERCICIO: notas de 3 alumnos en 2 exámenes (matriz) -----
notas = np.array([[5, 7],          # alumno 0
                  [8, 9],          # alumno 1
                  [4, 6]])         # alumno 2
print(notas.mean(axis=1))          # [6.  8.5 5. ]  media de cada alumno (por filas)
print(notas.mean(axis=0))          # [5.66666667 7.33333333]  media de cada examen
print(notas[notas >= 5])           # [5 7 8 9 6]   las notas aprobadas
```
