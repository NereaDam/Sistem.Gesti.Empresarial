```python
#Hacer un programa que pida los valores de los dos catetos 
#de un triángulo rectángulo, y calcule el valor de la hipotenusa. 
#Utilizar la librería math y la función sqrt: math.sqrt(x). 
#Antes de usarlo hay que importar math. Buscar ejemplos.
```

```python
import math
```

```python
cateto1=(int(input("Introduce la longitud de un cateto")))
cateto2=(int(input("Introduce la longitud del otro cateto")))
#calculo de la hipotenusa: math.hypto (math.hypot(cateto1,cateto2))

hipotenusa=math.hypot(cateto1,cateto2)
print("La longitud de la hipotesusa con math.hypot es: ", hipotenusa)
#tambien de puede calcular con math.sqrt (math.sqrt.(cateto1**2 + cateto2**2))
hipotenusa=math.sqrt(cateto1**2 + cateto2**2)
print("La longitud de la hipotesusa con math.srqt es: ", hipotenusa)
```

```python
#Un alumno desea saber cual será su calificación final en la materia
#de Algoritmos. Dicha calificación se compone de los siguientes 
#porcentajes:

#55% del promedio de sus tres calificaciones parciales.
#30% de la calificación del examen final. 
#15% de la calificación de un trabajo final.
```

```python
examen1=(int(input("Introduce la nota del primer examen")))
examen2=(int(input("Introduce la nota del primer examen")))
examen3=(int(input("Introduce la nota del primer examen")))
examen_final=(int(input("Introduce la nota del examen final")))
trabajo_final=(int(input("Introduce la nota del trabajo final")))

media_tres_examenes=((examen1+examen2+examen3)/3)
media_final_tres_examenes=media_tres_examenes*0.55
media_examen_final=examen_final*0.30
media_trabajo_final=trabajo_final*0.15

print("La media de tus 3 examenes es: ", media_final_tres_examenes)
print("La media del examen final es: ", media_examen_final)
print("La media del trabajo final es: ",  media_trabajo_final)
nota_final=(media_final_tres_examenes+media_examen_final+media_trabajo_final)
print("La nota final es: ", nota_final)
```
