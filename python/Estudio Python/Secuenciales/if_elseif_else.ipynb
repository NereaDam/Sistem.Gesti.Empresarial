```python
nota= float(input("Introduce nota"))

if nota<5:
    print("Suspenso, tu nota es", nota, "y es menor que 5")
elif nota<7:
    print("Aprobado, tu nota es", nota)
elif nota<9:
    print("Notable, tu nota es:", nota)
else:
    print("Sobresaliente, tu nota es: ", nota)
```
