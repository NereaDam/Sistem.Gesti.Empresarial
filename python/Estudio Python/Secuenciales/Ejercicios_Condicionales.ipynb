```python
#Hacer un programa que pida el precio y calcule el total. 
#Si el total supera los 100 euros, el programa ha de descontar 15€.
```

```python
precio=float(input("Introduce el precio"))

if precio>100:
    print("Enhorabuena! su compra supera los 100€ y se le hace un descuento de 15€")
    precio=precio-15

print("El total a pagar es:", precio)
```

```python
#Hacer un programa que introduciendo un número del 1 al 7, 
#diga si corresponde a lunes, martes, miércoles, ... 
#Si el número no está comprendido entre 1 y 7, ha de mostrar un mensaje 
#avisando que no puede decir el día.
```

```python
dia=int(input("Introduce en digito el dia de la semana"))

if dia<1:
    print("El numero introducido es menor a 1")
elif dia>7:
    print("El numero introducido es mayor a 7")
elif dia==1:
    print("LUNES")
elif dia==2:
    print("MARTES")
elif dia==3:
    print("MIERCOLES")
elif dia==4:
    print("JUEVES")
elif dia==5:
    print("VIERNES")
elif dia==6:
    print("SABADO")
else:
    print("DOMINGO")
```

```python
#Hacer un programa que pida dos números por el teclado, 
#si el primer número es divisible por el segundo, 
#los ha de sumar, sino, los ha de multiplicar.

numero1=int(input("inserta el primer numero"))
numero2=int(input("inserta el segundo numero"))

if numero1%numero2==0:
    suma= numero1+numero2
    print(numero1, "es diviidble por", numero2, ".La suma es ", suma)
else:
    multiplicacion=numero1*numero2
    print(numero1,"no es divisible por", numero2,".La multiplicacion es", multiplicacion)
```

```python
#Hacer un programa que pida dos números y que después los imprima en orden ascendente.
```

```python
numero1=int(input("Inserta el primer numero"))
numero2=int(input("Inserta el segundo numero"))

if numero1<numero2:
    print(numero1, "-",numero2)
else:
    print(numero2,"-", numero1)
```

```python
#Dados dos números, dar la relación entre ellos. 
#Señalar el mayor, el menor o especificar si son iguales
```

```python
numero1=int(input("Inserta el primer numero"))
numero2=int(input("Inserta el segundo numero"))

if numero1>numero2:
    print(numero1,"es mayor que", numero2)
elif numero1==numero2:
    print(numero1,"es igual a ", numero2)
else:
    print(numero2, "es mayor que ", numero1)
```
