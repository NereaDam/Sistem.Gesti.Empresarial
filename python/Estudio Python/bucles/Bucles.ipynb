```python
#While
```

```python
repeticiones=int(input("Introduce el numero de repeticiones "))
contador=1

while contador<=repeticiones:
    print(contador)
    contador+=1
print("Fin")
```

```python
#Bucles anidados
```

```python
x=1
while x <= 5:
    y=1
    while y <= 5:
        print(x," - ",y)
        y+=1
    x+=1
```

```python
#FOR
```

```python
#en el parentesis, el 1 seria desde donde empieza, como contador=0
#el 101, indica hasta donde 
#(en este caso es como un < restrictivo, no llega al 101, se queda en 100)
```

```python
for i in range (1,101):
    if i%2==0:
        print(i);
```

```python
#En etse for metemos 3 parametros 
#el primero indica desde donde empieza, en este caso en 2
#el segundo parametro indica hasta donde, en este caso hasta 100
#y el tercero indica que se cuentan de 2 en dos,los pares
```

```python
for i in range (2, 101,2):
    print(i)
```

```python
m=int(input("Introduce desde que numero contar"))
n=int(input("Introduce hasta que numero contar"))

for i in range (m,n):
    print(i)
```
