```python
#Hacer un programa que pida dos números por el teclado N y M, y si M > N 
#muestre todos los valores comprendidos entre estos dos números.
```

```python
m=int(input("Introduce un numero"))
n=int(input("Introduce otro numero"))

if m > n:
    while n<m:
        print(n)
        n+=1
        #se pone n=n+1, porque en cada vuelta n incrementa hasta llegar a m
print("Fin")
    
```
